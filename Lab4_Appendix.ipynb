```python
from collections import Counter, defaultdict
from statistics import mean

LAST_NAME = "Acapulco"
SEED_NUM = "5"
FAVORITE_ARTIST = "Takanashi Kiara"

SEED = int(SEED_NUM)
ARTIST_VALUES = [ord(char) for char in FAVORITE_ARTIST if char.isalpha()]

def generated_value(index, offset=0):
    char_value = ARTIST_VALUES[index % len(ARTIST_VALUES)]
    return SEED + char_value + offset
```

```python
def engineering_test_data():
    print("\n" + "=" * 60)
    print("EXERCISE 1: ENGINEERING TEST DATA ORGANIZER")
    print("=" * 60)

    sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
    categories = [
        "TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"
    ]

    records = []

    for i, source in enumerate(sources): 
        for j, category in enumerate(categories):
            value = generated_value(i + j, i + j)

            if category == "TEMPERATURE":
                value = 20 + value % 20
            elif category == "VOLTAGE":
                value = 1 + (value % 120) / 10
            elif category == "CURRENT":
                value = 1 + value % 20
            elif category == "PRESSURE":
                value = 90 + value % 30
            else:
                value = 10 + value % 50
            
            records.append({
                "source": source,
                "category": category,
                "value": round(value, 2)
            })

    records.append(records[0].copy())

    records.append({
        "source": sources[0],
        "category": "VOLTAGE",
        "value": 9.0
    })
    records.append({
        "source": sources[1],
        "category": "CURRENT",
        "value": 5.0
    })

    print("\n1. GENERATED TEST DATA")
    for record in records:
        print(record)
    
    grouped = defaultdict(list)
    source_counts = Counter()
    category_counts = Counter()
    record_counts = Counter()

    for recorf in records:
        grouped[record["source"]].append(record)
        source_counts[record["source"]] += 1
        category_counts[record["category"]] += 1
        key = (record["source"], record["category"])
        record_counts[key] += 1
    
    print("\n2. ORGANIZED DATA BY SOURCE")
    for source, items in sorted(grouped.items()):
        print(f"\n{source}:")
        for item in items:
            print(f" {item['category']}: {item['value']}")
        
    print("\n3. REPEATED RECORDS")
    repeated = {
        key: count for key, count in record_counts.items()
        if count > 1
    }
    print(repeated if repeated else "No repeated records")

    print("\n4. DISTINCT SOURCES AND CATEGORIES")
    print("Distinct sources:", len(source_counts))
    print("Distinct categories:", len(category_counts))
    print("Source counts:", dict(source_counts))
    print("Category counts:", dict(category_counts))

    print("\n5. NUMERICAL SUMMARIES NY CATEGORY")
    for category in categories:
        values = [
            r["value"] for r in records
            if r["category"] == category
        ]
        print(
            f"{category}: count={len(values)}, "
            f"min={min(values)}, max={max(values)}, "
            f"average={mean(values):.2f}"
        )
    
    print("\n6. ORDERED RESULTS")
    for record in sorted(
        records, key=lambda r: (r["category"], r["source"])
    ):
        print(record)
    
    print("\n7. FINAL ENGINEERING SUMMARY")
    print("Total records:", len(records))
    print("Unique sources:", len(source_counts))
    print("Unique categories:", len(category_counts))
    print("Repeated source/category combinations:", len(repeated))

print("Starting Exercise 1...")
engineering_test_data()
```

```python

students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

records = []

for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        score = 50 + generated_value(i + j, i * 3 + j) % 51
        records.append({
            "student": student,
            "subject": subject,
            "score": score
        })

records.append(records[0].copy())
records.append({
    "student": students[0],
    "subject": "PHYSICS",
    "score": None
})

def classify(score):
    if score is None:
        return "INCOMPLETE"
    if score >= 90:
        return "EXCELLENT"
    if score >= 80:
        return "VERY GOOD"
    if score >= 75:
        return "PASSED"
    return "FAILED"

print("1. STUDENT RECORDS")
for r in records:
    print(r, "->", classify(r["score"]))

counts = Counter((r["student"], r["subject"]) for r in records)

print("\n2. DUPLICATE RECORDS")
for key, count in counts.items():
    if count > 1:
        print(key, ":", count, "records")

print("\n3. STUDENT AVERAGES")
for student in students:
    scores = [
        r["score"] for r in records
        if r["student"] == student and r["score"] is not None
    ]
    print(student, ":", round(mean(scores), 2))

print("\n4. SUBJECT AVERAGES")
for subject in subjects:
    scores = [
        r["score"] for r in records
        if r["subject"] == subject and r["score"] is not None
    ]
    print(subject, ":", round(mean(scores), 2))

print("\n5. FINAL SUMMARY")
print("Total records:", len(records))
print("Valid scores:", sum(r["score"] is not None for r in records))
print("Incomplete scores:", sum(r["score"] is None for r in records))
print("Duplicate student-subject pairs:",
      sum(count > 1 for count in counts.values()))
```

```python
from collections import Counter

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
prices = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 250,
    "HEADPHONES": 500,
    "CALCULATOR": 350
}

transactions = []

for i, customer in enumerate(customers):
    for j, product in enumerate(prices):
        quantity = 1 + generated_value(i + j, i) % 5
        transactions.append({
            "customer": customer,
            "product": list(prices)[j],
            "quantity": quantity,
            "price": prices[list(prices)[j]]
        })

transactions.append(transactions[0].copy())

print("1. TRANSACTION RECORDS")
for t in transactions:
    t["total"] = t["quantity"] * t["price"]
    print(t)

print("\n2. CUSTOMER TOTALS")
for customer in customers:
    total = sum(t["total"] for t in transactions if t["customer"] == customer)
    print(customer, ": PHP", total)

print("\n3. PRODUCT SALES")
for product in prices:
    quantity = sum(t["quantity"] for t in transactions if t["product"] == product)
    sales = sum(t["total"] for t in transactions if t["product"] == product)
    print(product, "- Quantity:", quantity, "| Sales: PHP", sales)

counts = Counter((t["customer"], t["product"]) for t in transactions)

print("\n4. FINAL SUMMARY")
print("Total transactions:", len(transactions))
print("Unique customers:", len(customers))
print("Unique products:", len(prices))
print("Total sales: PHP", sum(t["total"] for t in transactions))
print("Duplicate customer-product pairs:", 
    sum(count > 1 for count in counts.values()))
```
